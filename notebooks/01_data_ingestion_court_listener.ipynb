```python
import os
import requests

# Running the workflow for the Examples (Same as "Run workflow" in GitHub UI)

# The Github PAT is stored as a Databricks Secret
github_token = dbutils.secrets.get(scope="dev_keys_and_tokens", key="GITHUB_PAT_2026")

# Targetting the specific workflow (api_data_examples.yaml)
url = f"https://api.github.com/repos/abdoulaity/ai-privacy-case-scanner-databricks/actions/workflows/api_data_examples.yaml/dispatches"
    
# JSON media type specific to GitHub's API
headers = {
    "Authorization": f"Bearer {github_token}",
    "Accept": "application/vnd.github+json"
}

# Running the workflow on the Main Branch
payload = {
    "ref": "main"  
}

# Responding only if it succeeds
response = requests.post(url, headers=headers, json=payload)
print("Status code:", response.status_code)

# Only parse JSON if the request failed
if response.status_code == 204:
    print("Success! Workflow triggered successfully.")
else:
    print("Error response:", response.json())
```
